```python
# 1부터 100까지의 짝수의 합을 ramge()의 증가값을 이용하여 구하세요.(if 문을 쓰지 말 것)
sum = 0

for i in range(2, 101, 2):
    sum = sum + i

print(sum)
```

```python
# 1부터 100까지의 짝수의 합을 ramge()의 증가값을 이용하여 구하세요.(if 문을 사용할 것)
sum = 0

for i in range(1, 101):
    if i % 2 == 0:
        sum = sum + i

print(sum)
```

```python
# 1부터 1000까지의 짝수의합을 range()의 증가값을 이용하여 구하세요.(if 문을 쓰지 말 것)
n = int(input("해당 구구단의 정수를 입력하세요:"))
for i in range(1,10):
    print(n, "x", i, "=", n*i)
```

```python
# 1부터 100까지 합을 출력하시오.
sum = 0
i = 1

while True:
    sum += i
    i += 1
    if i > 100:
        break

print(sum)
```

```python
# 양의 정수를 계속 입력받아 합계를 구하다가 0을 입력하면 반복을 끝내고 합계를 출력하세요.
sum = 0

while True:
    n = int(input("양의 정수를 입력하세요:"))
    if n == 0:
        break
    sum += n
print(sum)
```

```python
#1부터 20까지의 수 중 짝수만 담은 리스트를 리스트 내포로 만들어 출력하세요.
evens = []

for i in range(1, 21):
    if i % 2 == 0:
        evens.append(i)

print(evens)
```
